# Week 13 - Object Oriented Programming (OOP)

Python offers a whole bunch of different data types. For example the "string" method. We can define a string, and then apply methods to it (ie string.lower(), string.split(' '), etc.). But sometimes we may want to define a new "data type" with custom methods. This is possible with object-oriented programming (OOP) in which we can create a custom data type (called a "class") and then create objects related to this class (called "objects").

We will cover:
- Classes and objects
- Attributes
- Methods
- Inheritance
- Encapsulation

Becuase this is "Intermediate Programming for **Humanists**" we will imagine we are building this in the context of an art gallery.

### Classes, Objects, and Attributes

* Class: A new data type
* Object: A variable of this new data type
* Attribute: Properties of the data type (can be applied to all objects of class, or each instance/object of the class)


```python
# Basic class definition

class Artwork:
    # Class attribute
    beautiful=True

    # This is the "constructor" function. It is run each time a new object of this class is created
    def __init__(self, title, artist, year):
        # We now define attributes. The following are "instance attributes"
        self.title = title  
        self.artist = artist
        self.year = year
    
    def describe(self):
        return f"'{self.title}' by {self.artist} ({self.year})"

# Create an object
art1 = Artwork("Starry Night", "Vincent van Gogh", 1889)

print(art1.beautiful)
print(art1.describe())
```

## Methods and Behaviour

Methods define behaviour for objects.

```python
class Gallery:
    def __init__(self, name):
        self.name = name
        self.collection = []
    
    def add_artwork(self, artwork):
        self.collection.append(artwork)
    
    def show_collection(self):
        for art in self.collection:
            print(art.describe())

# Create gallery and add artworks
gallery = Gallery("Modern Art Gallery")
gallery.add_artwork(art1)

gallery.show_collection()
```

## Inheritance

We can create specialized types of artworks using inheritance.

```python
class Painting(Artwork):
    def __init__(self, title, artist, year, medium):
        super().__init__(title, artist, year)
        self.medium = medium
    
    def describe(self):
        return f"'{self.title}' by {self.artist} ({self.year}) - Medium: {self.medium}"

painting1 = Painting("The Persistence of Memory", "Salvador Dalí", 1931, "Oil on canvas")

print(painting1.describe())
```

## Encapsulation

Encapsulation restricts direct access to some data.

By putting an underscore in front of an attribute or method it is understood that it should only be used internally within the class definition, and not outside from the rest of the code.

Note: In python this is just convention - no way to enforce that these attributes cannot be accessed from the outside.

```python
class Collector:
    def __init__(self, name, budget):
        self.name = name
        self._budget = budget  # protected attribute
    
    def get_budget(self):
        return self._budget
    
    def buy_art(self, price):
        if price <= self._budget:
            self._budget -= price
            print(f"Purchase successful! Remaining budget: {self._budget}")
        else:
            print("Not enough budget.")

collector = Collector("Alice", 10000)
collector.buy_art(3000)
collector.buy_art(8000)

print(collector._budget)

# Note: If using "__budget" with two underscores this goes from "protected" to "private". It can no longer be printed
# (in fact, it can still be accessed by "collector._Collector__budget", but this should not be done)
```

Copy the "Collector" class from above. We will update the "buy_art" method so that instead of taking a price it now takes an object of the "artwork" class.
```python
class Collector:
    def __init__(self, name, budget):
        self.name = name
        self.budget = budget
    
    def buy_art(self, artwork):
        if artwork.price <= self.budget and not artwork.sold:
            self.budget -= artwork.price
            artwork.sold = True
            print(f"{self.name} bought {artwork.title}")
```
