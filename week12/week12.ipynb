# Week 12 - Object Oriented Programming (OOP)

This notebook introduces core object-oriented programming (OOP) concepts in Python using an art gallery scenario.

We will cover:
- Classes and objects
- Attributes
- Methods
- Inheritance
- Encapsulation


```python
# Basic class definition

class Artwork:
    def __init__(self, title, artist, year):
        self.title = title
        self.artist = artist
        self.year = year
    
    def describe(self):
        return f"'{self.title}' by {self.artist} ({self.year})"

# Create an object
art1 = Artwork("Starry Night", "Vincent van Gogh", 1889)

print(art1.describe())
```

## Methods and Behaviour

Methods define behaviour for objects.

```python
class Gallery:
    def __init__(self, name):
        self.name = name
        self.collection = []
    
    def add_artwork(self, artwork):
        self.collection.append(artwork)
    
    def show_collection(self):
        for art in self.collection:
            print(art.describe())

# Create gallery and add artworks
gallery = Gallery("Modern Art Gallery")
gallery.add_artwork(art1)

gallery.show_collection()
```

## Inheritance

We can create specialized types of artworks using inheritance.

```python
class Painting(Artwork):
    def __init__(self, title, artist, year, medium):
        super().__init__(title, artist, year)
        self.medium = medium
    
    def describe(self):
        return f"'{self.title}' by {self.artist} ({self.year}) - Medium: {self.medium}"

painting1 = Painting("The Persistence of Memory", "Salvador Dalí", 1931, "Oil on canvas")

print(painting1.describe())
```

## Encapsulation

Encapsulation restricts direct access to some data.

```python
class Collector:
    def __init__(self, name, budget):
        self.name = name
        self.__budget = budget  # protected attribute
    
    def get_budget(self):
        return self.__budget
    
    def buy_art(self, price):
        if price <= self.__budget:
            self.__budget -= price
            print(f"Purchase successful! Remaining budget: {self.__budget}")
        else:
            print("Not enough budget.")

collector = Collector("Alice", 10000)
collector.buy_art(3000)
collector.buy_art(8000)
```
