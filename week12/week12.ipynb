# Week 12 - GitHub

This notebook is a step-by-step guide for GitHub/Visual Studio integration.


1) Make a GitHub account

2) On the left side-bar click "Extensions"

3) Search for and download "GitHub Pull Requests"

4) On the bottom of the left side-bar click "accounts" then "Sign in with GitHub to use GitHub Pull Requests"

5) Sign into your GitHub and authorize everything

6) On the left side-bar click "Source Control" and "Open Folder" then create a new folder

7) Create a python file (.py) in the folder. You can do this in visual studio if you install the "python" extension

8) Back in "Source Control" click "Publish to GitHub". Cloose "Publish to public repository"

9) If prompted on the bottom right to periodically run "Git Fetch" say "Yes" (this regularly checks for new changes in the code at the origin.

# Some terminology

### Repo (Repository) = A bunch of code, generally many files

### Origin = the server (your GitHub repo online)

### Local = your computer (the version of the repo on your computer)

### Branch = One version of the code

10) Go online and in your GitHub account you now have your first Repo! It will ask if you want to add a ReadMe. This is a description of your project to help others use it. Create one. You can use markdown in the ReadMe file.
11) Back on Visual Studio go to Soure control. There is a button to Sync. Click it. Now your ReadMe file is also on your local repo.
12) How do we run python code if we're not in a notebook? Click on "Terminal" on the top toolbar, then "New Terminal". You can check the folder you're currently in with "ls". And you can run a python file with "python FILENAME.py".
13) When you modify code, go to Source Control and write a commit message (a description of what has been changed). Then you can "commit" the changes (they are ready to be ) "push" the code form your local repo to the "origin".

# Branching

When we want to develop the code, we may want to leave the main code safe and functional until we are finished developping. To do this we "branch" away from the main code, work within this new branch, and then rejoin the main master branch when we are finished (through something called a "pull request" or "PR")

14) On the bottom left you can see the branch you are currently in (the "master" branch). Click on this and creat a new branch named "dev".
15) Publish the branch via source control.
16) Add some code and sync it.
17) Now go on GitHub and you can see there are multiple branches.
18) You can create a Pull Request through either Visual Studio or on GitHub.
    * Visual Studio: Click on the "GitHub" button in the left tool bar. Then the pull request option is a small icon next to "Pull Requests"
    * GitHub: Top menu click on "Pull Requests" tap (3rd from left). Then click "Compare & pull request" (or new pull request)
19) We'll use the "GitHub" option. Add a title & description. You can also assign reviewers, labels, projects, milestones, etc. You can also create a "Draft Pull Request" which cannot be merged until the "Draft" property is removed.
20) When you are ready, you can merge!
21) Now the dev branch and master branch are the same. You can delete the dev branch, or keep developping on it.
22) Back in Visual Studio, synchronize and everything should be the same.
