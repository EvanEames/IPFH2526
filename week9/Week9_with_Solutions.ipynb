```python
def understanding1(word, input_list):
    if word in input_list:
        input_list.remove(word)
    else:
        input_list.append(word)
    return input_list

understanding1("cat", ["dog", "fish", "cow"])
```

```python
def understanding2(input_numb):
    if input_numb > 5:
        input_numb = input_numb + 5
    if input_numb > 8:
        input_numb = input_numb - 3
    if input_numb > 10:
        input_numb = input_numb + 1
    if input_numb > 20:
        input_numb = input_numb // 5
    return input_numb

understanding2(15)
```

```python
def understanding3(code, dict_code):
    sentence = []
    for numb in code:
        if numb*2 in dict_code:
            sentence.append(dict_code[numb*2])
        else:
            sentence.append(" ")
    return sentence

input_dictionary = {2:'a', 4:'b', 6:'c', 8:'d', 10:'e',
                    12:'f', 14:'g', 16:'h', 18:'i', 20:'j',
                    22:'k', 24:'l', 26:'m', 28:'n', 30:'o',
                    32:'p', 34:'q', 36:'r', 38:'s', 40:'t',
                    42:'u', 44:'v', 46:'w', 48:'x', 50:'y',
                    52:'z'}

understanding3([9,28,12,15,22,5,29,16,25,10,15,11,5], input_dictionary)
```

```python
def understanding4(code, last_name):
    result_dictionary = dict()
    last_name = set(last_name)
    last_name = list(last_name)
    print(last_name)
    for numb in code:
        if numb < len(last_name):
            result_dictionary[last_name[numb]] = numb
        else:
            new_letter = last_name.pop()
            last_name.insert(1000, new_letter)
    return result_dictionary

understanding4([5,10,4,3,7], ["s","a","r","t","i","n","i"])
```
