## Understanding

The goal of the understanding part of algorithms is that you are able to predict what the output will be without running the code by recognizing the different python methods that are used.

For example, given:

```python
x = "hello world"
y = "aeiou"
count = 0
for char in x:
    if char in y:
        count += 1
```

What is the value of `count` at the end?

Or a more complex example, given this function:

```python
def my_function(digits_list):
    result = []
    second_digit_list = []
    for n in digits_list:
        second_digit_list.append(n)
    for n in digits_list:
        if n > len(second_digit_list):
            result.append(n)
        else:
            second_digit_list.pop()
    return result
```
            

What does it return if I run it as `my_function([0,9,3,1,5,2])`?

Acting like a computer, you should try to dissect the function like this:


First for loop: iterates over the input list and gives its values to a second list called `second_digit_list`
Current status after the first for loop
* `second_digit_list`: is basically a copy of the first list, so it's value at the start is `[0,9,3,1,5,2]`
* `result`: is still empty, no value was added to it

Second for loop: iterates over the input list and according to it's values it adds something to the `result` list or it changes the `second_digit_list`
* first iteration:  `n = 0`. `n` is not more than `6` (which the current length of `second_digit_list`). So the algorithm takes out the last element of `second_digit_list`
* `result` is still empty, `second_digit_list` is now `[0,9,3,1,5]`
* second iteration `n = 9`. `n` is more than `5` (which is the current length of `second_digit_list`). So the algorithm adds `9` to `result`.
* `result` is now `[9]`, `second_digit_list` is still `[0,9,3,1,5]`
* third iteration `n = 3`. `n` is not more than `5` (which is the current length of `second_digit_list`). So the algorithm takes out the last element of `second_digit_list`
* `result` is still `[9]`, `second_digit_list` is now `[0,9,3,1]`
* fourth iteration: `n = 1`. `n` is not more than `4` (which is the current length of `second_digit_list`). So the algorithm takes out the last element of `second_digit_list`
* `result` is still `[9]`, `second_digit_list` is now `[0,9,3]`
* fifth iteration: `n = 5`. `n` is more than `3` (which is the current length of `second_digit_list`). So the algorithm adds `5` to `result`.
*  `result` is now `[9, 5]`, `second_digit_list` is still `[0,9,3]`
* sixth iteration:  `n = 2`. `n` is not more than `3` (which is the current length of `second_digit_list`). So the algorithm takes out the last element of `second_digit_list`
* `result` is still `[9, 5]`, `second_digit_list` is now `[0,9]`
* The algorithm returns `result`, which is `[9, 5]`

````python
# Just as a proof of concept, we run the algorithm traditionally

def my_function(digits_list):
    result = []
    second_digit_list = []
    for n in digits_list:
        second_digit_list.append(n)
    for n in digits_list:
        if n > len(second_digit_list):
            result.append(n)
        else:
            second_digit_list.pop()
    return result

my_function([0,9,3,1,5,2])
````

# 8.2

Write a function `calculate_frequency(text)` that takes a string of text as input and returns a dictionary where the keys are the unique characters in the text and the values are the frequencies of each character. Ignore any whitespace characters. For example, if the input is `"hello world"`, the function should return `{'h': 1, 'e': 1, 'l': 3, 'o': 2, 'w': 1, 'r': 1, 'd': 1}`.

````python
def calculate_frequency(text):
    text = text.replace(' ','')
    unique_chars = {}
    for char in text:
        if char in unique_chars.keys():
            unique_chars[char] += 1
        else:
            unique_chars[char] = 1
    return unique_chars


calculate_frequency("hello world")
````

# 8.3
Write a function `merge_dicts(dicts)` that takes a list of dictionaries as input and returns a single dictionary that merges all the dictionaries in the list. If there are any common keys among the dictionaries, the values should be combined into a list. For example, if the input is `[{1: 'a', 2: 'b'}, {2: 'c', 3: 'd'}, {4: 'e'}]`, the function should return `{1: 'a', 2: ['b', 'c'], 3: 'd', 4: 'e'}`.

````python
def merge_dicts(dicts):
    master_dict = {}
    for dict in dicts:
        for key in dict.keys():
            if key in master_dict.keys():
                master_dict[key].append(dict[key])
            else:
                master_dict[key] = [dict[key]]
    return master_dict

merge_dicts([{1: 'a', 2: 'b'}, {2: 'c', 3: 'd'}, {4: 'e'}])
````

Write a function `reverse_string(string)` that takes a string as input and returns the reversed version of the string. For example, if the input is `"hello"`, the function should return `"olleh"`.

````python
def reverse_string(string):
    string = list(string)
    new_string = []
    while len(string) > 0:
        new_string.append(string.pop())  # Repeatedly pop letters out of the string list and into the new string
    return ''.join(new_string)

# Pro way of doing it:
def makhamed_reverse_string(string):
    return string[::-1]

makhamed_reverse_string("hello what'sup")
````

# 8.6

Write a function `word_count(sentence)` that takes a sentence as input and returns a dictionary containing the count of each word in the sentence. The keys of the dictionary should be the unique words in the sentence, and the values should be the corresponding counts. Ignore any punctuation and treat uppercase and lowercase letters as the same word. For example, if the input is `"I love to code, and code is fun!"`, the function should return `{"i": 1, "love": 1, "to": 1, "code": 2, "and": 1, "is": 1, "fun": 1}`.

````python
# Quite similar to 8.2, except we need to first switch the string to be lowercase and remove punctuation, and then we are counting words instead of letters

def word_count(sentence):
    sentence = sentence.lower().replace(',','').replace('!','')
    dict = {}
    for word in sentence.split(' '):
        if word in dict.keys():
            dict[word] += 1
        else:
            dict[word] = 1
    return dict

word_count("I love to code, and code is fun!")
````

# 8.7


Write a function `update_csv(filename, search_column, search_value, update_value)` that takes a CSV filename, a search column, a search value, an update column, and an update value as input. The function should search for the row that matches the search criteria in the CSV file using `DictReader`, update the value in the specified update column, and save the changes back to the CSV file. Make sure to handle cases where multiple rows match the search criteria.

## example CSV

```
Name,Age,City
John,25,New York
Alice,30,Los Angeles
Bob,35,Chicago
```

````python
# Quite awkward with the built in python stuff:

import csv

def update_csv(filename, search_column, search_value, update_column, update_value):
    with open(filename, 'r') as input_file:
        dict_list = []
        reader = csv.DictReader(input_file)
        for row in reader:  # We iterate over the dictionaries (rows) and update the values as needed:
            if row[search_column] == search_value:
                row[update_column] = update_value
            dict_list.append(row)  # For each iteration we add the dictionaries to our dict_list

    # Now we write everything out. Note that we have to provide the dictionary keys, which we can do by just taking them from the first dictionary (dict_list[0]) and switching them to a list
    with open("new_"+filename, 'w') as output_file:
        writer = csv.DictWriter(output_file, fieldnames=list(dict_list[0].keys()))
        writer.writeheader()
        for dict in dict_list:
            writer.writerow(dict)
    return None

update_csv('example.csv', 'City', 'Los Angeles', 'City', 'Mogadishu')
````

````python
# Way cleaner with pandas:

import pandas as pd

def update_csv_with_pandas(filename, search_column, search_value, update_column, update_value):
    df = pd.read_csv(filename)
    df.loc[df[search_column] == search_value, update_column] = update_value
    df.to_csv('new_'+filename, index=False)  # Without index=Fales the indeces (0,1,2,3,4,5,...) will be printed as a first row
    
update_csv_with_pandas('example.csv', 'City', 'Los Angeles', 'City', 'Mogadishu')
````

# 8.8

Write a function merge_csv_files(input_files, output_file, field) that takes a list of input CSV filenames and an output CSV filename as input. The function should read the input CSV files using DictReader, merge them into a single CSV file, and write the merged data to the output CSV file. The merged CSV file should contain all the columns from the input files, and the rows should be concatenated in the order of the input files.

Example input_files:

```
Name,Age
John,25
Alice,30
Carol,24
```

```
Name,City
Bob,Chicago
Carol,Houston

```

Merged output:

```
Name,Age,City
John,25,
Alice,30,
Bob,,Chicago
Carol,24,Houston
```

````python
import csv

def merge_csv_files(input_files, output_file):
    # We want to keep track of both all dictionaries and all unique keys
    dict_list = []
    keys_list = []
    for file in input_files:
        with open(file, 'r') as input_file:
            reader = csv.DictReader(input_file)
            for row in reader:
                dict_list.append(row)
        # Before moving to the next file, remember the dictionary keys (column names) of the current file
        keys_list = keys_list + list(row.keys())
    keys_list = list(set(keys_list))  # This remove duplicate column names (present in both files)

    # Now write out all our dictionaries collected in the list from all files, specifying the keys_list (keys from all files)
    with open(output_file, 'w') as output_file:
        writer = csv.DictWriter(output_file, fieldnames=keys_list)
        writer.writeheader()
        for dict in dict_list:
            writer.writerow(dict)
    return None

merge_csv_files(["csv1.csv","csv2.csv"], "merged.csv")
````

# Exercise 8.9

Write a function `analyze_sales_data(input_file, product_name)` that takes an input CSV filename as input. The CSV file contains sales data with columns like 'Date', 'Product', 'Price', and 'Quantity'. The function should read the CSV file using `pd.read_csv` and perform the following operations:

Filter the data to include only the sales records for a specific product (e.g., 'Product A').
Calculate the total revenues (Price * Quantity) for each day and put them in a list as tuples (date, revenue)
Create a new column, called month and assigning the correct month to each column based on the 'Date' column.
Group the data by month and compute the average revenue for each month. Add the average revenues to a list as tuples (month, revenue)
Return both lists.

## Example CSV

```
Date,Product,Price,Quantity
2023-01-15,Product A,10.99,5
2023-01-16,Product B,15.99,10
2023-01-17,Product A,10.99,8
2023-02-15,Product A,10.99,3
2023-02-16,Product A,10.99,6
2023-02-17,Product B,15.99,12
```

````python
import pandas as pd

def analyze_sales_data(input_file, product_name):
    df = pd.read_csv(input_file)
    df = df[df['Product'] == product_name]  # Filter out all products except those we are interested in
    df["Revenue"] = df["Price"]*df["Quantity"]  # Make a new column "Revenue" which is equal to Price*Quantity
    revenue_list = list(zip(df['Date'], df["Revenue"]))  # "zip" combines the "Date" and "Revenue" columns into a list of tuples, as the problem asks for
    df["Month"] = [date[1] for date in df["Date"].str.split('-')]  # When we split the dates by their hyphens we get a list for each row which has the format date = ["yyyy","mm","dd"]. So then date[1] gives us only the month, which we save to a new column
    df = df.groupby("Month").mean(numeric_only=True)  # Group by month and take the mean (ignoring non numeric columns)
    average_revenues = list(zip(df.index, df['Revenue']))  # The grouped dataframe has "Month" as an index, so we can get it with df.index. We zip this the Revenue column and hand them back
    return revenue_list, average_revenues
````

 Write a function `analyze_heritage_data(input_file)` that takes the input CSV filename as input. The function should read the CSV file using `pd.read_csv` and perform the following operations:

Find the top 5 countries with the highest number of cultural heritage sites, store the top 5 in a list ordered from the first to the last.

````python
# This question is a bit odd. Mainly it asks for two things (top countries for # of sites, and average visitors), but it's not clear what we have to return. I just returned both
# It's also silly to assume that visitor # has been constant since the pyramids were made 4500 years ago! But oh well

from collections import Counter
import re

def analyze_heritage_data(input_file):
    df = pd.read_csv(input_file)

    top_countries = dict(Counter(df["Country"]))  # This gives us a nice dictionary containing the countries as keys and the # of sites as values
    sorted_top_countries = sorted(top_countries.items(), key=lambda item: item[1], reverse=True)  # We can sort the countries by their values (item[1] for each item in the dictionary) and save it into a list of tuples. Reverse=True makes it so that the list is in descending order
    top_countries = [k for k, v in sorted_top_countries]  # For each tuple pair (country, count) in our sorted list of tuples we only need the first value (country). We save these into a normal list
    top_countries = top_countries[0:5]  # And take the first 5

    avg_visitors = []
    # With the zip function we can loop over the dataframe and use both the year AND the visitor count for each row we iterate on.
    for year, visitorCount in zip(df["Year"], df["VisitorCount"]):
        year_parsed = re.findall(r'\d+', year)  # We can use regex to return a list of all numbers we find in the "year column" for each row
        # Now we need to deal with each possibility case-by case. Kinda annoying, but there's not really a better way to do it
        if "AD" in year or len(year) <= 4:
            age = 2023 - int(year_parsed[0])
        elif "BC" in year:
            age = 2023 + int(year_parsed[0])
        elif "century" in year:  # If we are using the "century" unit we need to multiply by 100 (ie 5th century = 5*100 = 500 years)
            if "BC" in year:
                age = 2023 + int(year_parsed[0])*100
            else:
                age = 2023 - int(year_parsed[0])*100
        elif "-" in year:  # If we have a date range we can just take the mean
            average_year = [int(year_parsed[0]),int(year_parsed[1])].mean()
            if "BC" in year:
                age = 2023 + average_year
            else:
                age = 2023 - average_year
            
        avg_visitors.append(visitorCount//age)  # Now we average the visitors per year (as said, kind of silly)
    avg_visitors_dict = {k:v for k,v in zip(df["Site"],avg_visitors)}  # This is the quickest way I could find to combine a dataframe column (site) and a list (avg_visitors) into a dictionary
    
    return top_countries, avg_visitors_dict

analyze_heritage_data("Exercise9.csv")
````
